```python
n = 10
a = 1
b = 1
for i in range(3, n):
    aux = a + b
    a = b
    b = aux
print(b)
```

```python
a = 1
b = 1
for i in range(3, n + 1):
    a, b = b, a + b
print(b)
```

```python
es_primo = True
d = 2
while d * d <= n and es_primo:
    if n % d == 0:
        es_primo = False
    d = d + 1
```
